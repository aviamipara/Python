```python
# Write a program for web scrapping using BeautifulSoup to scrape the following details from the given 
# link and make a data frame using that scraped data from the page in a given link.
# Link: https://www.politifact.com/factchecks
# You will find 30 news articles with fact checks on this page. You need to scrape the following details 
# from all the articles and store that in a data frame. Statement of News, Date of News, Source of News. 



import requests
from bs4 import BeautifulSoup

url="https://www.politifact.com/factchecks"
response=requests.get(url)
print(response.status_code)
response=response.text 
soup=BeautifulSoup(response,"html.parser") 
# print(soup.prettify())

```

```python
a=soup.find("div",class_="o-listicle__inner").find_all("li",class_="o-listicle__item")
news=[]
date=[]
source=[]
for i in a:
    news.append(i.find("div",class_="m-statement__quote").find("a",href=True).text.strip())
    d=i.find("footer",class_="m-statement__footer").text.strip()
    d1=d.split("•")
    date.append(d1[1])
    source.append(i.find("div",class_="m-statement__desc").text.strip())

import pandas as pd

df=pd.DataFrame({"News ":news , "Date ":date,"Source ":source})
df
```

```python
from bs4 import BeautifulSoup
import pandas as pd

file=open("web_scraping.html",encoding="utf-8")
response=file.read()
soup=BeautifulSoup(response,"html.parser")
# soup=soup.prettify()
```

```python

table = soup.find("table")
# print(table.prettify())

cols = table.find_all("th")
columns = []

for i in cols:
    col = i.get_text().strip()
    columns.append(col)

print(columns)

body = table.find("tbody")

trs = body.find_all("tr")

all_data = []

for i in range(len(trs)):
    tds = trs[i].find_all("td")

    li = []

    for j in range(len(tds)):
        txt = tds[j].get_text().strip()
        li.append(txt)
    all_data.append(li)


df = pd.DataFrame(all_data , columns = columns)
df
    
```
